# Synthetic MRI denoising demo
Create a reproducible Shepp–Logan phantom, add magnitude Rician noise, and compare methods. Synthetic results do not establish clinical utility.

```python
from skimage.data import shepp_logan_phantom
from skimage.transform import resize
from mri_denoise.noise import add_rician_noise
from mri_denoise.evaluate import evaluate_methods
import matplotlib.pyplot as plt
import numpy as np
clean = resize(shepp_logan_phantom(), (256, 256), anti_aliasing=True).astype(np.float32)
noisy = add_rician_noise(clean, sigma=0.05, seed=7)
rows = evaluate_methods(clean, noisy, sigma=0.05)
rows
```

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(clean, cmap='gray', vmin=0, vmax=1); axes[0].set_title('Reference')
axes[1].imshow(noisy, cmap='gray', vmin=0, vmax=1); axes[1].set_title('Rician noisy')
for ax in axes: ax.axis('off')
plt.tight_layout()
```
